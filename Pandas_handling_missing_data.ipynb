# Pandas handling missing data

```python
import pandas as pd
import numpy as np

nfl_data = pd.read_csv("train.csv")

np.random.seed(0) 
```

```python
nfl_data.head()
```

### Counting Missing Data

```python
missing_values_count = nfl_data.isnull().sum()

missing_values_count
```

```python
total_cells = np.prod(nfl_data.shape)
total_missing = missing_values_count.sum()

percent_missing = (total_missing/total_cells) * 100
print(percent_missing)
```

### Determining the Cause of Missing Data

```python
missing_values_count[10:11]
```

### Dropping Missing Values

```python
nfl_data.dropna()
```

```python
columns_with_na_dropped = nfl_data.dropna(axis=1)
columns_with_na_dropped.head()
```

```python
print("Кількість стовпців : %d \n" % nfl_data.shape[1])
print("Кількість видалених стовпців: %d" % columns_with_na_dropped.shape[1])
```

### Imputing Missing Values with Zeros

```python
nfl_data.fillna(0)
```

### Creating a Data Subset 

```python
subset_nfl_data = nfl_data[['Age', 'Cabin', 'Embarked']].head(10)
subset_nfl_data
```

```python
subset_nfl_data.fillna(100)
```

```python
nfl_data.bfill(axis=0).fillna(0)
```
