```python
book.iloc[:4]
```

```python
book.iloc[-4:]
```

```python
book.loc[0, 'author']
```

```python
book.loc[:,['rating','author','year of publication']]
```

```python
book.loc[:5,['rating','author','year of publication']]
```

```python
book.loc[1:100:11,['rating','author','year of publication']]
```

```python
book.set_index("book title")
```

```python
pmean=book['book price'].mean()
print("Середня ціна =",pmean)
book.loc[book['book price']>pmean]
```

```python
pmean=book['book price'].mean()
print("Середня ціна =",pmean)
book.loc[(book['book price']<pmean) & (book['year of publication']==2018)]
```

```python
book.loc[(book.genre=="Memoir") | (book.rating>=4.8)]
```

```python
book.loc[book.author.isin(['Craig Smith'])]
```

```python
book.loc[book.rating.isin([4.4,4.5])]
```

```python
book.loc[book.author.isnull()]
```
