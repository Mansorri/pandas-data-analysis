```python
review_IMDB_Rating_mean = reviews['IMDB Rating'].mean()
reviews['IMDB Rating'].map(lambda p: p - review_IMDB_Rating_mean)
```

```python
us_gross_mean = reviews['US Gross'].mean()

def remean_us_gross(row):
    row['US Gross'] = row['US Gross'] - us_gross_mean
    row['Worldwide Gross'] = row['Worldwide Gross'] + row['Worldwide Gross'] * 0.2
    return row
    
reviews = reviews.apply(remean_us_gross, axis='columns')
reviews.head()
```
